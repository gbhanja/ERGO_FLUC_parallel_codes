```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os


downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "TC_Kac_OP_fock_bosonic_omega.npz")

data = np.load(filename)

# extract arrays
ω0_arr2 = data["ω0"]
tau2 = data["tau"]
E_ergo2 = data["Eerg"]
E_var2 = data["variance"]
E_ratio2 = data["ratio"]
E_B2 = data["Eb"]


# derived quantities
Power2 = E_B2 / tau2


F_ratio2 = np.sqrt(E_var2) / E_ergo2
FE_ratio2 = np.sqrt(E_var2) / E_B2
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
##########################################################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

N = 50

E_max_arr1 = ω0_arr1 * N
E_max_arr2 = ω0_arr2 * N

# Numerical data
x1 = E_ergo1 / E_max_arr1
y1 = F_ratio1

x2 = E_ergo2 / E_max_arr2
y2 = F_ratio2

# Domain for theoretical bounds
xb = np.linspace(0.01, 1.0, 400)

# Upper bounds
bound_finite = np.sqrt(1 / xb**2 - 1)
bound_largeN = np.sqrt(1 / xb - 1)

# --------------------------------------------------------
# Figure and axes
# --------------------------------------------------------

fig, ax = plt.subplots(figsize=(14, 6), dpi=400)

# --------------------------------------------------------
# Common normalization
# --------------------------------------------------------

norm = Normalize(
    vmin=min(ω0_arr1.min(), ω0_arr2.min()),
    vmax=max(ω0_arr1.max(), ω0_arr2.max())
)

# --------------------------------------------------------
# Numerical results
# --------------------------------------------------------

sc1 = ax.scatter(
    x1,
    y1,
    c=ω0_arr1,
    cmap='Blues',
    norm=norm,
    marker='o',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Distinguishable battery'
)

sc2 = ax.scatter(
    x2,
    y2,
    c=ω0_arr2,
    cmap='Reds',
    norm=norm,
    marker='s',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Indistinguishable battery'
)

# --------------------------------------------------------
# Theoretical bounds
# --------------------------------------------------------

ax.plot(
    xb,
    bound_finite,
    'r-',
    lw=2,
    label=r'General upper bound: $\sqrt{({\mathcal{E}_b/E_{\max}})^{-2}-1}$'
)

ax.plot(
    xb,
    bound_largeN,
    'g--',
    lw=2,
    label=r'Positive-work upper bound: $\sqrt{({\mathcal{E}_b/E_{\max}})^{-1}-1}$'
)

# --------------------------------------------------------
# Axis limits and labels
# --------------------------------------------------------

ax.set_xlim(-0.02, 1.0)
ax.set_ylim(0, 4)

ax.set_xlabel(
    r'$\mathcal{E}_b/E_{\max}$',
    fontsize=13
)

ax.set_ylabel(
    r'$\sigma_{\mathcal{E}_b}/\langle \mathcal{E}_b\rangle$',
    fontsize=13
)

ax.set_title(
    f"Relative ergotropy fluctuations and theoretical upper bounds of {model} model for N={N}",
    fontsize=11
)

# --------------------------------------------------------
# Colorbar for distinguishable battery
# --------------------------------------------------------

sm1 = ScalarMappable(
    norm=norm,
    cmap='Blues'
)

sm1.set_array([])

cbar1 = fig.colorbar(
    sm1,
    ax=ax,
    pad=0.02
)

cbar1.set_label(
    r'$\omega_B$ — distinguishable',
    fontsize=12
)

cbar1.set_ticks(
    [1, 5, 10, 15]
)

# --------------------------------------------------------
# Colorbar for indistinguishable battery
# --------------------------------------------------------

sm2 = ScalarMappable(
    norm=norm,
    cmap='Reds'
)

sm2.set_array([])

cbar2 = fig.colorbar(
    sm2,
    ax=ax,
    pad=0.10
)

cbar2.set_label(
    r'$\omega_B$ — indistinguishable',
    fontsize=12
)

cbar2.set_ticks(
    [1, 5, 10, 15]
)

# --------------------------------------------------------
# Legend
# --------------------------------------------------------

ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
```
