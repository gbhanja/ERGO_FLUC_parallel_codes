```python
##############################################################
# log(Relative ergotropy fluctuations) vs log(N)
# for omega_0 = 1.0
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100, 200, 300, 400
])


# ============================================================
# Maximum battery energy
# ============================================================

E_max1 = omega0 * N_arr1
E_max2 = omega0 * N_arr2


# ============================================================
# Normalized ergotropy
# ============================================================

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(
    N_arr1,
    N_selected
)

mask2 = np.isin(
    N_arr2,
    N_selected
)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]
x1_sel = x1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]
x2_sel = x2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]
x1_sel = x1_sel[i1]

i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]
x2_sel = x2_sel[i2]


# ============================================================
# Keep only positive values for logarithm
# ============================================================

valid1 = (
    (N1 > 0) &
    (F1 > 0)
)

valid2 = (
    (N2 > 0) &
    (F2 > 0)
)


# ============================================================
# Explicit logarithmic transformation
# ============================================================

logN1 = np.log(N1[valid1])
logF1 = np.log(F1[valid1])

logN2 = np.log(N2[valid2])
logF2 = np.log(F2[valid2])


# ============================================================
# Theoretical bounds
# ============================================================

valid_bound = (
    (x1_sel > 0) &
    (x1_sel <= 1)
)

N_bound = N1[valid_bound]
x_bound = x1_sel[valid_bound]


general_bound = np.sqrt(
    1.0 / x_bound**2 - 1.0
)

positive_work_bound = np.sqrt(
    1.0 / x_bound - 1.0
)


# ============================================================
# Keep only positive bounds for logarithm
# ============================================================

valid_general = general_bound > 0
valid_positive = positive_work_bound > 0


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots(
    figsize=(10, 7),
    dpi=300
)


# ============================================================
# Distinguishable battery
# ============================================================

ax.plot(
    logN1,
    logF1,
    'o-',
    ms=7,
    lw=2.0,
    label='Distinguishable battery',
    zorder=4
)


# ============================================================
# Indistinguishable battery
# ============================================================

ax.plot(
    logN2,
    logF2,
    's--',
    ms=7,
    lw=2.0,
    label='Indistinguishable battery',
    zorder=4
)


# ============================================================
# General upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_general]),
    np.log(general_bound[valid_general]),
    color='black',
    linestyle='-',
    linewidth=2.0,
    label='General upper bound',
    zorder=2
)


# ============================================================
# Positive-work upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_positive]),
    np.log(positive_work_bound[valid_positive]),
    color='black',
    linestyle='--',
    linewidth=2.0,
    label='Positive-work upper bound',
    zorder=2
)


# ============================================================
# Axis labels
# ============================================================

ax.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)

ax.set_ylabel(
    r'$\log\!\left('
    r'\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle'
    r'\right)$',
    fontsize=15
)


# ============================================================
# Title
# ============================================================

ax.set_title(
    rf'Log-relative ergotropy fluctuations vs $\log(N)$ '
    rf'of the {model} model for $\omega_0={omega0}$',
    fontsize=14
)


# ============================================================
# X ticks 
# ============================================================

logN_ticks = np.log(N_selected)

ax.set_xticks(logN_ticks)

ax.set_xticklabels(
    [f'{np.log(N):.1f}' for N in N_selected],
    fontsize=11
)


# ============================================================
# Tick parameters
# ============================================================

ax.tick_params(
    axis='y',
    labelsize=11
)



# ============================================================
# Legend
# ============================================================

ax.legend(
    fontsize=10,
    frameon=True,
    loc='best'
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Difference in log(relative ergotropy fluctuations)
# between distinguishable and indistinguishable batteries
# as a function of log(N)
# for omega_0 = 1.0
##############################################################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100, 200, 300, 400
])


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(
    N_arr1,
    N_selected
)

mask2 = np.isin(
    N_arr2,
    N_selected
)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]


i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]


# ============================================================
# Find common N values
# ============================================================

common_N = np.intersect1d(
    N1,
    N2
)


# ============================================================
# Extract F values at common N
# ============================================================

F_dist = np.array([
    F1[N1 == N][0]
    for N in common_N
])

F_indist = np.array([
    F2[N2 == N][0]
    for N in common_N
])


# ============================================================
# Keep only positive fluctuations
# ============================================================

valid = (
    (common_N > 0) &
    (F_dist > 0) &
    (F_indist > 0)
)

N_common = common_N[valid]

F_dist = F_dist[valid]
F_indist = F_indist[valid]


# ============================================================
# Logarithms
# ============================================================

logN = np.log(N_common)

logF_dist = np.log(F_dist)

logF_indist = np.log(F_indist)


# ============================================================
# Difference in logarithmic fluctuations
# ============================================================

delta_logF = (
    logF_dist - logF_indist
)


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots(
    figsize=(9, 7),
    dpi=500
)


# ============================================================
# Main curve
# ============================================================

ax.plot(
    logN,
    delta_logF,
    'o-',
    color='black',
    markersize=8,
    linewidth=2,
    label=r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$'
)


# ============================================================
# Zero reference
# ============================================================

ax.axhline(
    0,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=r'$\Delta\log \mathcal{F}=0$'
)


# ============================================================
# Axis labels
# ============================================================

ax.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)

ax.set_ylabel(
    r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist}),$'
    r'$\mathcal{F} = \sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle $',
    fontsize=15
)


# ============================================================
# Title
# ============================================================

ax.set_title(
    rf'Difference in logarithmic relative ergotropy '
    rf'fluctuations of the {model} model for $\omega_0={omega0}$',
    fontsize=14
)


# ============================================================
# X-axis ticks
# ============================================================

logN_ticks = np.log(N_selected)

ax.set_xticks(logN_ticks)

ax.set_xticklabels(
    [f'{np.log(N):.1f}' for N in N_selected],
    fontsize=11
)

# ============================================================
# Tick parameters
# ============================================================

ax.tick_params(
    axis='y',
    labelsize=11
)


# ============================================================
# Legend
# ============================================================

ax.legend(
    fontsize=10,
    frameon=True,
    loc='best'
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Vertical subplots for omega_0 = 1.0
#
# Top:
# log(relative ergotropy fluctuation) vs log(N)
#
# Bottom:
# difference between distinguishable and indistinguishable
# logarithmic relative fluctuations vs log(N)
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100, 200, 300, 400
])


# ============================================================
# Maximum battery energy
# ============================================================

E_max1 = omega0 * N_arr1
E_max2 = omega0 * N_arr2


# ============================================================
# Normalized ergotropy
# ============================================================

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(N_arr1, N_selected)
mask2 = np.isin(N_arr2, N_selected)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]
x1_sel = x1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]
x2_sel = x2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]
x1_sel = x1_sel[i1]

i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]
x2_sel = x2_sel[i2]


# ============================================================
# Keep only positive values for logarithms
# ============================================================

valid1 = (
    (N1 > 0) &
    (F1 > 0)
)

valid2 = (
    (N2 > 0) &
    (F2 > 0)
)


# ============================================================
# Logarithmic quantities
# ============================================================

logN1 = np.log(N1[valid1])
logF1 = np.log(F1[valid1])

logN2 = np.log(N2[valid2])
logF2 = np.log(F2[valid2])


# ============================================================
# Common N values for difference plot
# ============================================================

common_N = np.intersect1d(N1, N2)

F_dist = np.array([
    F1[N1 == N][0]
    for N in common_N
])

F_indist = np.array([
    F2[N2 == N][0]
    for N in common_N
])


valid_diff = (
    (common_N > 0) &
    (F_dist > 0) &
    (F_indist > 0)
)

N_common = common_N[valid_diff]

F_dist = F_dist[valid_diff]
F_indist = F_indist[valid_diff]


# ============================================================
# Logarithmic difference
# ============================================================

logN_diff = np.log(N_common)

logF_dist = np.log(F_dist)
logF_indist = np.log(F_indist)

delta_logF = logF_dist - logF_indist


# ============================================================
# Theoretical bounds
# ============================================================

valid_bound = (
    (x1_sel > 0) &
    (x1_sel <= 1)
)

N_bound = N1[valid_bound]
x_bound = x1_sel[valid_bound]


general_bound = np.sqrt(
    1.0 / x_bound**2 - 1.0
)

positive_work_bound = np.sqrt(
    1.0 / x_bound - 1.0
)


valid_general = general_bound > 0
valid_positive = positive_work_bound > 0


# ============================================================
# Create vertical subplots
# ============================================================

fig, axes = plt.subplots(
    2,
    1,
    figsize=(10, 12),
    dpi=500,
    sharex=True,
    gridspec_kw={'height_ratios': [1, 1]}
)

ax1, ax2 = axes


# ============================================================
# TOP PLOT
# log(relative fluctuation) vs log(N)
# ============================================================

ax1.plot(
    logN1,
    logF1,
    'o-',
    ms=7,
    lw=2.0,
    label='Distinguishable battery',
    zorder=4
)

ax1.plot(
    logN2,
    logF2,
    's--',
    ms=7,
    lw=2.0,
    label='Indistinguishable battery',
    zorder=4
)


# ------------------------------------------------------------
# General upper bound
# ------------------------------------------------------------

ax1.plot(
    np.log(N_bound[valid_general]),
    np.log(general_bound[valid_general]),
    color='black',
    linestyle='-',
    linewidth=2.0,
    label='General upper bound',
    zorder=2
)


# ------------------------------------------------------------
# Positive-work upper bound
# ------------------------------------------------------------

ax1.plot(
    np.log(N_bound[valid_positive]),
    np.log(positive_work_bound[valid_positive]),
    color='black',
    linestyle='--',
    linewidth=2.0,
    label='Positive-work upper bound',
    zorder=2
)


# ------------------------------------------------------------
# Top plot title
# ------------------------------------------------------------

ax1.set_title(
    rf'Log-relative ergotropy fluctuations vs $\log(N)$ ',
    fontsize=14
)


# ------------------------------------------------------------
# Top y-axis
# ------------------------------------------------------------

ax1.set_ylabel(
    r'$ \mathcal{F} = \sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle $',
    fontsize=15
)


# ------------------------------------------------------------
# Top legend
# ------------------------------------------------------------

ax1.legend(
    fontsize=9,
    frameon=True,
    loc='best'
)


# ============================================================
# BOTTOM PLOT
# Difference between the logarithmic fluctuations
# ============================================================

ax2.plot(
    logN_diff,
    delta_logF,
    'o-',
    color='black',
    ms=7,
    lw=2.0,
    label=r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$'
)


# ------------------------------------------------------------
# Zero reference
# ------------------------------------------------------------

ax2.axhline(
    0,
    color='red',
    linestyle='--',
    linewidth=1.5,
    label=r'$\Delta\log(\mathcal{F})=0$'
)

# ------------------------------------------------------------
# Bottom plot title
# ------------------------------------------------------------

ax2.set_title(
    r'Difference in logarithmic fluctuations',
    fontsize=14
)

# ------------------------------------------------------------
# Bottom y-axis
# ------------------------------------------------------------

ax2.set_ylabel(
    r'$\log(\mathcal{F}_{\rm dist})-\log(\mathcal{F}_{\rm indist})$',
    fontsize=15
)


# ============================================================
# Common x-axis
# ============================================================

ax2.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)


# ============================================================
# X-axis ticks
# ============================================================

logN_ticks = np.log(N_selected)

for ax in axes:

    ax.set_xticks(logN_ticks)

    ax.set_xticklabels(
        [f'{np.log(N):.1f}' for N in N_selected],
        fontsize=11
    )

    ax.tick_params(
        axis='y',
        labelsize=11
    )


# ============================================================
# Bottom legend
# ============================================================

ax2.legend(
    fontsize=9,
    frameon=True,
    loc='best'
)


# ============================================================
# Overall title
# ============================================================

fig.suptitle(
    rf'Relative ergotropy fluctuations in the {model} model for $\omega_0={omega0}$',
    fontsize=17,
    y=0.98
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout(
    rect=[0.05, 0.04, 0.98, 0.95]
)

plt.show()
```
