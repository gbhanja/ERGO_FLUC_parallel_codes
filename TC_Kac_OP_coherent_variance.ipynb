```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os


downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "TC_Kac_OP_coherent_bosonic.npz")

data = np.load(filename)

# extract arrays
N_arr2 = data["N"]
tau2 = data["tau"]
E_ergo2 = data["Eerg"]
E_var2 = data["variance"]
E_ratio2 = data["ratio"]
E_B2 = data["Eb"]


# derived quantities
Power2 = E_B2 / tau2


F_ratio2 = np.sqrt(E_var2) / E_ergo2
FE_ratio2 = np.sqrt(E_var2) / E_B2
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
##########################################################

ω0 = 1.0
E_max1 = ω0 * N_arr1         # Largest battery energy
E_max2 = ω0 * N_arr2         # Largest battery energy


# Numerical data
x1 = E_ergo1 / E_max1
y1 = F_ratio1

x2 = E_ergo2 / E_max2
y2 = F_ratio2

# Domain for theoretical bounds
xb = np.linspace(0.01, 1.0, 400)

# Upper bounds 
bound_finite = np.sqrt(1/xb**2 - 1)      # General upper bound
bound_largeN = np.sqrt(1/xb - 1)         # Positive-work upper bound

plt.figure(figsize=(14,6), dpi=500)

# Numerical results

from matplotlib.colors import Normalize

norm = Normalize(
    vmin=min(N_arr1.min(), N_arr2.min()),
    vmax=max(N_arr1.max(), N_arr2.max())
)

sc1 = plt.scatter(
    x1, y1,
    c=N_arr1,
    cmap='Blues',
    norm=norm,
    marker='o',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Distinguishable battery'
)

sc2 = plt.scatter(
    x2, y2,
    c=N_arr2,
    cmap='Reds',
    norm=norm,
    marker='s',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Indistinguishable battery'
)

# General upper bound
plt.plot(
    xb,
    bound_finite,
    'r-',
    lw=2,
    label=r'General upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-2} -1}$'
)

# Positive-work upper bound
plt.plot(
    xb,
    bound_largeN,
    'g--',
    lw=2,
    label=r'Positive-work upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-1} -1}$'
)

plt.xlim(0.01,1.0)
plt.ylim(0,4)

plt.xlabel(r'$\mathcal{E}_b/E_{\max}$', fontsize=13)
plt.ylabel(r'$\sigma_{\mathcal{E}_b}/\ \langle \mathcal{E}_b \rangle$', fontsize=13)

plt.title(f"Relative ergotropy fluctuations and theoretical upper bounds of {model} model", fontsize=11)

# Two colorbars

cbar2 = plt.colorbar(sc2, pad=0.02)
cbar2.set_label(r'$N (Indistinguishable)$', color='red')

cbar1 = plt.colorbar(sc1, pad=0.10)
cbar1.set_label(r'$N (Distinguishable)$', color='blue')

plt.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
# Fixed N = 100, varying omega_0
##########################################################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable


# ========================================================
# Fixed system size
# ========================================================

N_fixed = 100


# ========================================================
# omega_0 values
# ========================================================

omega_values = np.arange(1.0, 15.0, 0.1)


# ========================================================
# Select N = 100
# ========================================================

idx1 = N_arr1 == N_fixed
idx2 = N_arr2 == N_fixed


if not np.any(idx1):
    raise ValueError("N=100 not found in N_arr1.")

if not np.any(idx2):
    raise ValueError("N=100 not found in N_arr2.")


# Numerical data at N_fixed
E1_Nfixed = E_ergo1[idx1]
F1_Nfixed = F_ratio1[idx1]

E2_Nfixed = E_ergo2[idx2]
F2_Nfixed = F_ratio2[idx2]


# ========================================================
# Theoretical bounds
# ========================================================

xb = np.linspace(0.01, 1.0, 500)

bound_finite = np.sqrt(
    1 / xb**2 - 1
)

bound_largeN = np.sqrt(
    1 / xb - 1
)


# ========================================================
# Figure
# ========================================================

fig, ax = plt.subplots(
    figsize=(14, 8),
    dpi=500
)


# ========================================================
# Continuous color mapping for omega_0
# ========================================================

norm = Normalize(
    vmin=omega_values.min(),
    vmax=omega_values.max()
)

cmap = plt.cm.viridis


# ========================================================
# Plot numerical data
# ========================================================

for omega0 in omega_values:

    color = cmap(norm(omega0))


    # ----------------------------------------------------
    # Maximum battery energy
    # ----------------------------------------------------

    E_max1 = omega0 * N_fixed
    E_max2 = omega0 * N_fixed


    # ----------------------------------------------------
    # Normalized ergotropy
    # ----------------------------------------------------

    x1 = E1_Nfixed / E_max1
    x2 = E2_Nfixed / E_max2


    # ----------------------------------------------------
    # Distinguishable battery
    # ----------------------------------------------------

    ax.scatter(
        x1,
        F1_Nfixed,
        color=color,
        marker='o',
        s=70,
        edgecolors='k',
        linewidth=0.5
    )


    # ----------------------------------------------------
    # Indistinguishable battery
    # ----------------------------------------------------

    ax.scatter(
        x2,
        F2_Nfixed,
        color=color,
        marker='s',
        s=70,
        edgecolors='k',
        linewidth=0.5
    )


# ========================================================
# Theoretical upper bounds
# ========================================================

ax.plot(
    xb,
    bound_finite,
    color='black',
    linestyle='-',
    linewidth=2,
    label='General upper bound'
)

ax.plot(
    xb,
    bound_largeN,
    color='black',
    linestyle='--',
    linewidth=2,
    label='Positive-work upper bound'
)


# ========================================================
# Axis limits
# ========================================================

ax.set_xlim(0.01, 1.0)
ax.set_ylim(0.0, 2.5)


# ========================================================
# Axis labels
# ========================================================

ax.set_xlabel(
    r'$\mathcal{E}_b/E_{\max}$',
    fontsize=14
)

ax.set_ylabel(
    r'$\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle$',
    fontsize=14
)


# ========================================================
# Title
# ========================================================

ax.set_title(
    r'Relative ergotropy fluctuations for fixed '
    rf'$N={N_fixed}$ and varying $\omega_B$ of {model} model',
    fontsize=13
)


# ========================================================
# Battery-type legend
# ========================================================

battery_legend = [

    Line2D(
        [0],
        [0],
        marker='o',
        linestyle='None',
        color='gray',
        markeredgecolor='black',
        markersize=8,
        label='Distinguishable battery'
    ),

    Line2D(
        [0],
        [0],
        marker='s',
        linestyle='None',
        color='gray',
        markeredgecolor='black',
        markersize=8,
        label='Indistinguishable battery'
    )
]


# ========================================================
# Theoretical-bound legend
# ========================================================

bound_legend = [

    Line2D(
        [0],
        [0],
        color='black',
        linestyle='-',
        linewidth=2,
        label='General upper bound'
    ),

    Line2D(
        [0],
        [0],
        color='black',
        linestyle='--',
        linewidth=2,
        label='Positive-work upper bound'
    )
]


# ========================================================
# Combined legend
# ========================================================

ax.legend(
    handles=battery_legend + bound_legend,
    fontsize=9,
    loc='best',
    ncol=2,
    frameon=True
)


# ========================================================
# Colorbar for omega_0
# ========================================================

sm = ScalarMappable(
    norm=norm,
    cmap=cmap
)

sm.set_array([])

cbar = fig.colorbar(
    sm,
    ax=ax,
    pad=0.02
)

cbar.set_label(
    r'$\omega_B$',
    fontsize=14
)


# ========================================================
# Colorbar ticks
# ========================================================

cbar.set_ticks(
    [1, 5, 10, 15]
)


# ========================================================
# Layout
# ========================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Difference in log(relative ergotropy fluctuations)
# between distinguishable and indistinguishable batteries
# as a function of log(N)
# for omega_0 = 1, 5, 10, 15
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# omega_0 values
# ============================================================

omega_values = [1.0, 5.0, 10.0, 15.0]


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100, 200, 300, 400
])


# ============================================================
# Create figure
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(15, 11),
    sharex=True,
    sharey=True
)

axes = axes.flatten()


# ============================================================
# Loop over omega_0
# ============================================================

for ax, omega0 in zip(axes, omega_values):

    # --------------------------------------------------------
    # Select requested N values
    # --------------------------------------------------------

    mask1 = np.isin(
        N_arr1,
        N_selected
    )

    mask2 = np.isin(
        N_arr2,
        N_selected
    )


    N1 = N_arr1[mask1]
    F1 = F_ratio1[mask1]

    N2 = N_arr2[mask2]
    F2 = F_ratio2[mask2]


    # --------------------------------------------------------
    # Sort according to N
    # --------------------------------------------------------

    i1 = np.argsort(N1)

    N1 = N1[i1]
    F1 = F1[i1]


    i2 = np.argsort(N2)

    N2 = N2[i2]
    F2 = F2[i2]


    # --------------------------------------------------------
    # Find common N values
    # --------------------------------------------------------

    common_N = np.intersect1d(
        N1,
        N2
    )


    # --------------------------------------------------------
    # Extract F values at common N
    # --------------------------------------------------------

    F_dist = np.array([
        F1[N1 == N][0]
        for N in common_N
    ])

    F_indist = np.array([
        F2[N2 == N][0]
        for N in common_N
    ])


    # --------------------------------------------------------
    # Keep only positive fluctuations
    # --------------------------------------------------------

    valid = (
        (F_dist > 0) &
        (F_indist > 0) &
        (common_N > 0)
    )


    N_common = common_N[valid]
    F_dist = F_dist[valid]
    F_indist = F_indist[valid]


    # ========================================================
    # Logarithms
    # ========================================================

    logN = np.log(N_common)

    logF_dist = np.log(F_dist)

    logF_indist = np.log(F_indist)


    # ========================================================
    # Difference between the two logarithmic fluctuations
    # ========================================================

    delta_logF = (
        logF_dist - logF_indist
    )


    # ========================================================
    # Plot difference
    # ========================================================

    ax.plot(
        logN,
        delta_logF,
        'o-',
        color='black',
        markersize=7,
        linewidth=2,
        label=r'$\log(F_{\rm dist})-\log(F_{\rm indist})$'
    )


    # ========================================================
    # Zero reference line
    # ========================================================

    ax.axhline(
        0,
        color='red',
        linestyle='--',
        linewidth=1.5
    )


    # ========================================================
    # Panel title
    # ========================================================

    ax.set_title(
        rf'$\omega_0 = {omega0}$',
        fontsize=14,
        pad=10
    )


    # ========================================================
    # Grid
    # ========================================================

    ax.grid(
        True,
        alpha=0.25
    )


    # ========================================================
    # Tick parameters
    # ========================================================

    ax.tick_params(
        axis='both',
        labelsize=11
    )


# ============================================================
# Common x-axis
# ============================================================

fig.supxlabel(
    r'$\log(N)$',
    fontsize=17,
    y=0.045
)


# ============================================================
# Common y-axis
# ============================================================

fig.supylabel(
    r'$\log(F_{\rm dist})-\log(F_{\rm indist})$',
    fontsize=17,
    x=0.025
)


# ============================================================
# Overall title
# ============================================================

fig.suptitle(
    r'Difference in logarithmic relative ergotropy fluctuations',
    fontsize=17,
    y=0.985
)


# ============================================================
# Interpretation legend
# ============================================================

from matplotlib.lines import Line2D

legend_elements = [

    Line2D(
        [0],
        [0],
        color='black',
        marker='o',
        linewidth=2,
        markersize=7,
        label=r'$\log(F_{\rm dist})-\log(F_{\rm indist})$'
    ),

    Line2D(
        [0],
        [0],
        color='red',
        linestyle='--',
        linewidth=1.5,
        label=r'Equal fluctuations: $\Delta\log F=0$'
    )
]


fig.legend(
    handles=legend_elements,
    loc='upper center',
    ncol=2,
    fontsize=10,
    frameon=True,
    bbox_to_anchor=(0.5, 0.935)
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout(
    rect=[0.07, 0.08, 0.98, 0.88]
)

plt.show()
```

```python
##############################################################
# log(Relative ergotropy fluctuations) vs log(N)
# for omega_0 = 1.0
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Fixed omega_0
# ============================================================

omega0 = 1.0


# ============================================================
# Selected N values
# ============================================================

N_selected = np.array([
    2, 10, 50, 100, 200, 300, 400
])


# ============================================================
# Maximum battery energy
# ============================================================

E_max1 = omega0 * N_arr1
E_max2 = omega0 * N_arr2


# ============================================================
# Normalized ergotropy
# ============================================================

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# ============================================================
# Select requested N values
# ============================================================

mask1 = np.isin(
    N_arr1,
    N_selected
)

mask2 = np.isin(
    N_arr2,
    N_selected
)

N1 = N_arr1[mask1]
F1 = F_ratio1[mask1]
x1_sel = x1[mask1]

N2 = N_arr2[mask2]
F2 = F_ratio2[mask2]
x2_sel = x2[mask2]


# ============================================================
# Sort according to N
# ============================================================

i1 = np.argsort(N1)

N1 = N1[i1]
F1 = F1[i1]
x1_sel = x1_sel[i1]

i2 = np.argsort(N2)

N2 = N2[i2]
F2 = F2[i2]
x2_sel = x2_sel[i2]


# ============================================================
# Keep only positive values for logarithm
# ============================================================

valid1 = (
    (N1 > 0) &
    (F1 > 0)
)

valid2 = (
    (N2 > 0) &
    (F2 > 0)
)


# ============================================================
# Explicit logarithmic transformation
# ============================================================

logN1 = np.log(N1[valid1])
logF1 = np.log(F1[valid1])

logN2 = np.log(N2[valid2])
logF2 = np.log(F2[valid2])


# ============================================================
# Theoretical bounds
# ============================================================

valid_bound = (
    (x1_sel > 0) &
    (x1_sel <= 1)
)

N_bound = N1[valid_bound]
x_bound = x1_sel[valid_bound]


general_bound = np.sqrt(
    1.0 / x_bound**2 - 1.0
)

positive_work_bound = np.sqrt(
    1.0 / x_bound - 1.0
)


# ============================================================
# Keep only positive bounds for logarithm
# ============================================================

valid_general = general_bound > 0
valid_positive = positive_work_bound > 0


# ============================================================
# Figure
# ============================================================

fig, ax = plt.subplots(
    figsize=(10, 7),
    dpi=300
)


# ============================================================
# Distinguishable battery
# ============================================================

ax.plot(
    logN1,
    logF1,
    'o-',
    ms=7,
    lw=2.0,
    label='Distinguishable battery',
    zorder=4
)


# ============================================================
# Indistinguishable battery
# ============================================================

ax.plot(
    logN2,
    logF2,
    's--',
    ms=7,
    lw=2.0,
    label='Indistinguishable battery',
    zorder=4
)


# ============================================================
# General upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_general]),
    np.log(general_bound[valid_general]),
    color='black',
    linestyle='-',
    linewidth=2.0,
    label='General upper bound',
    zorder=2
)


# ============================================================
# Positive-work upper bound
# ============================================================

ax.plot(
    np.log(N_bound[valid_positive]),
    np.log(positive_work_bound[valid_positive]),
    color='black',
    linestyle='--',
    linewidth=2.0,
    label='Positive-work upper bound',
    zorder=2
)


# ============================================================
# Axis labels
# ============================================================

ax.set_xlabel(
    r'$\log(N)$',
    fontsize=15
)

ax.set_ylabel(
    r'$\log\!\left('
    r'\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle'
    r'\right)$',
    fontsize=15
)


# ============================================================
# Title
# ============================================================

ax.set_title(
    rf'Log-relative ergotropy fluctuations vs $\log(N)$ '
    rf'of the {model} model for $\omega_0={omega0}$',
    fontsize=14
)


# ============================================================
# X ticks 
# ============================================================

logN_ticks = np.log(N_selected)

ax.set_xticks(logN_ticks)

ax.set_xticklabels(
    [f'{np.log(N):.1f}' for N in N_selected],
    fontsize=11
)


# ============================================================
# Tick parameters
# ============================================================

ax.tick_params(
    axis='y',
    labelsize=11
)



# ============================================================
# Legend
# ============================================================

ax.legend(
    fontsize=10,
    frameon=True,
    loc='best'
)


# ============================================================
# Layout
# ============================================================

plt.tight_layout()

plt.show()
```
