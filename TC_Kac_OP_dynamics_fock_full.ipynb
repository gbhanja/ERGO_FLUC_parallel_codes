```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os

downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "TC_Kac_OP_dynamics_fock_full.npz")

data = np.load(filename)

# extract arrays
N_arr = data["N"]
tlist   = data["tlist"]
E_ergo = data["Eerg"]
E_B = data["Eb"]
```

```python
# ====================================
# Plot: Battery Energy Dynamics
# ====================================

plt.figure(figsize=(8, 6))

for i, N in enumerate(N_arr):
    plt.plot(tlist, E_B[i], lw=2, label=fr"$N={N}$")

plt.xlabel(r'$\tau$', fontsize=14)
plt.ylabel(r'$E_B/N$', fontsize=14)

plt.xlim(tlist[0], tlist[-1])
plt.title('Battery Energy Dynamics', fontsize=14)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
# ====================================
# Plot: Ergotropy Dynamics
# ====================================

plt.figure(figsize=(8,6))

for i, N in enumerate(N_arr):
    plt.plot(tlist, E_ergo[i], lw=2, label=fr"$N={N}$")

plt.xlabel(r'$\tau$', fontsize=14)
plt.ylabel(r'$\mathcal{E}_B/N$', fontsize=14)

plt.xlim(tlist[0], tlist[-1])
plt.title('Battery Ergotropy Dynamics', fontsize=14)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,8))

# Battery energy
for i, N in enumerate(N_arr):
    ax[0].plot(tlist, E_B[i], lw=2, label=fr"$N={N}$")

ax[0].set_xlabel(r'$\tau$', fontsize=14)
ax[0].set_ylabel(r'$E_B/N$', fontsize=14)
ax[0].legend(fontsize=10)


# Ergotropy
for i, N in enumerate(N_arr):
    ax[1].plot(tlist, E_ergo[i], lw=2, label=fr"$N={N}$")

ax[1].set_xlabel(r'$\tau$', fontsize=14)
ax[1].set_ylabel(r'$\mathcal{E}_B/N$', fontsize=14)
ax[1].legend(fontsize=10)
plt.suptitle('Battery Energy and Ergotropy Dynamics', fontsize=14)
plt.tight_layout()
plt.show()
```
