```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os

ω0 = 1.0

downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "Dicke_Kac_OP_fock_bosonic.npz")

data = np.load(filename)

# extract arrays
N_arr2 = data["N"]
tau2 = data["tau"]
E_ergo2 = data["Eerg"]
E_var2 = data["variance"]
E_ratio2 = data["ratio"]
E_B2 = data["Eb"]


# derived quantities
Power2 = E_B2 / tau2
E_max2 = ω0 * N_arr2         # Largest battery energy

F_ratio2 = np.sqrt(E_var2) / E_ergo2
FE_ratio2 = np.sqrt(E_var2) / E_B2
```

```python


for i in range(len(N_arr1)):
    print(
        f"N={N_arr1[i]}, "
        f"τ={tau1[i]:.6f}, "
        f"Eb={E_B1[i]:.6f}, "
        f"Ergotropy={E_ergo1[i]:.6f}, "
        f"Ratio={E_ratio1[i]:.6f}, "
        f"Variance={E_var1[i]:.10e},"
        f"Fluctuation w.r.t erg={F_ratio1[i]:.10e},"
        f"Fluctuation w.r.t EB={FE_ratio1[i]:.10e}"
    )
```

```python


for i in range(len(N_arr2)):
    print(
        f"N={N_arr2[i]}, "
        f"τ={tau2[i]:.6f}, "
        f"Eb={E_B2[i]:.6f}, "
        f"Ergotropy={E_ergo2[i]:.6f}, "
        f"Ratio={E_ratio2[i]:.6f}, "
        f"Variance={E_var2[i]:.10e},"
        f"Fluctuation w.r.t erg={F_ratio2[i]:.10e},"
        f"Fluctuation w.r.t EB={FE_ratio2[i]:.10e}"
    )
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
##########################################################

# Numerical data
x1 = E_ergo1 / E_max1
y1 = F_ratio1

x2 = E_ergo2 / E_max2
y2 = F_ratio2

# Domain for theoretical bounds
xb = np.linspace(0.01, 1.0, 400)

# Upper bounds 
bound_finite = np.sqrt(1/xb**2 - 1)      # General upper bound
bound_largeN = np.sqrt(1/xb - 1)         # Positive-work upper bound

plt.figure(figsize=(12,6))

# Numerical results

from matplotlib.colors import Normalize

norm = Normalize(
    vmin=min(N_arr1.min(), N_arr2.min()),
    vmax=max(N_arr1.max(), N_arr2.max())
)

sc1 = plt.scatter(
    x1, y1,
    c=N_arr1,
    cmap='Blues',
    norm=norm,
    marker='o',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Distinguishable battery'
)

sc2 = plt.scatter(
    x2, y2,
    c=N_arr2,
    cmap='Reds',
    norm=norm,
    marker='s',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Indistinguishable battery'
)

# General upper bound
plt.plot(
    xb,
    bound_finite,
    'r-',
    lw=2,
    label=r'General upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-2} -1}$'
)

# Positive-work upper bound
plt.plot(
    xb,
    bound_largeN,
    'g--',
    lw=2,
    label=r'Positive-work upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-1} -1}$'
)

plt.xlim(0.01,1.0)
plt.ylim(0,4)

plt.xlabel(r'$\mathcal{E}_b/E_{\max}$', fontsize=13)
plt.ylabel(r'$\sigma_{\mathcal{E}_b}/\ \langle \mathcal{E}_b \rangle$', fontsize=13)

plt.title("Relative ergotropy fluctuations and theoretical upper bounds of Dicke model", fontsize=11)

# Two colorbars

cbar2 = plt.colorbar(sc2, pad=0.02)
cbar2.set_label(r'$N (Indistinguishable)$', color='red')

cbar1 = plt.colorbar(sc1, pad=0.10)
cbar1.set_label(r'$N (Distinguishable)$', color='blue')

plt.legend(fontsize=10)

plt.tight_layout()
plt.show()
```
