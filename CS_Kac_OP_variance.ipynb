```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
import os

ω0 = 1.0

downloads = os.path.join(os.path.expanduser("~"), "Downloads")

filename = os.path.join(downloads, "CS_Kac_OP_bosonic.npz")

data = np.load(filename)

# extract arrays
N_arr2 = data["N"]
tau2 = data["tau"]
E_ergo2 = data["Eerg"]
E_var2 = data["variance"]
E_ratio2 = data["ratio"]
E_B2 = data["Eb"]


# derived quantities
Power2 = E_B2 / tau2
E_max2 = ω0 * N_arr2         # Largest battery energy

F_ratio2 = np.sqrt(E_var2) / E_ergo2
FE_ratio2 = np.sqrt(E_var2) / E_B2
```

```python


for i in range(len(N_arr2)):
    print(
        f"N={N_arr2[i]}, "
        f"τ={tau2[i]:.6f}, "
        f"Eb={E_B2[i]:.6f}, "
        f"Ergotropy={E_ergo2[i]:.6f}, "
        f"Ratio={E_ratio2[i]:.6f}, "
        f"Variance={E_var2[i]:.10e},"
        f"Fluctuation w.r.t erg={F_ratio2[i]:.10e},"
        f"Fluctuation w.r.t EB={FE_ratio2[i]:.10e}"
    )
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
##########################################################

# Numerical data
x1 = E_ergo1 / E_max1
y1 = F_ratio1

x2 = E_ergo2 / E_max2
y2 = F_ratio2

# Domain for theoretical bounds
xb = np.linspace(0.01, 1.0, 400)

# Upper bounds 
bound_finite = np.sqrt(1/xb**2 - 1)      # General upper bound
bound_largeN = np.sqrt(1/xb - 1)         # Positive-work upper bound

plt.figure(figsize=(14,6))

# Numerical results

from matplotlib.colors import Normalize

norm = Normalize(
    vmin=min(N_arr1.min(), N_arr2.min()),
    vmax=max(N_arr1.max(), N_arr2.max())
)

sc1 = plt.scatter(
    x1, y1,
    c=N_arr1,
    cmap='Blues',
    norm=norm,
    marker='o',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Distinguishable battery'
)

sc2 = plt.scatter(
    x2, y2,
    c=N_arr2,
    cmap='Reds',
    norm=norm,
    marker='s',
    s=45,
    edgecolors='k',
    linewidth=0.3,
    label='Indistinguishable battery'
)

# General upper bound
plt.plot(
    xb,
    bound_finite,
    'r-',
    lw=2,
    label=r'General upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-2} -1}$'
)

# Positive-work upper bound
plt.plot(
    xb,
    bound_largeN,
    'g--',
    lw=2,
    label=r'Positive-work upper bound: $\sqrt{({\mathcal{E}_b/E_{max}})^{-1} -1}$'
)

plt.xlim(0.01,1.0)
plt.ylim(0,4)

plt.xlabel(r'$\mathcal{E}_b/E_{\max}$', fontsize=13)
plt.ylabel(r'$\sigma_{\mathcal{E}_b}/\ \langle \mathcal{E}_b \rangle$', fontsize=13)

plt.title("Relative ergotropy fluctuations and theoretical upper bounds of Central Spin model", fontsize=11)

# Two colorbars

cbar2 = plt.colorbar(sc2, pad=0.02)
cbar2.set_label(r'$N (Indistinguishable)$', color='red')

cbar1 = plt.colorbar(sc1, pad=0.10)
cbar1.set_label(r'$N (Distinguishable)$', color='blue')

plt.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

```python
##########################################################
# Relative ergotropy fluctuations vs normalized ergotropy
# Selected N values
##########################################################

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D


# ========================================================
# Numerical data
# ========================================================

x1 = E_ergo1 / E_max1
y1 = F_ratio1

x2 = E_ergo2 / E_max2
y2 = F_ratio2


# ========================================================
# Selected N values
# ========================================================

N_selected = np.array([2, 10, 50, 100, 200, 300, 400])


# Keep only selected N values
mask1 = np.isin(N_arr1, N_selected)
mask2 = np.isin(N_arr2, N_selected)

x1_sel = x1[mask1]
y1_sel = y1[mask1]
N1_sel = N_arr1[mask1]

x2_sel = x2[mask2]
y2_sel = y2[mask2]
N2_sel = N_arr2[mask2]


# ========================================================
# Domain for theoretical bounds
# ========================================================

xb = np.linspace(0.01, 1.0, 400)


# General upper bound
bound_finite = np.sqrt(1 / xb**2 - 1)

# Positive-work upper bound
bound_largeN = np.sqrt(1 / xb - 1)


# ========================================================
# Figure
# ========================================================

plt.figure(figsize=(14, 7))


# ========================================================
# Colour assigned to each N
# ========================================================

colors = plt.cm.viridis(
    np.linspace(0, 1, len(N_selected))
)

N_colors = dict(zip(N_selected, colors))


# ========================================================
# Numerical results
#
# Same colour = same N
# Circle = distinguishable
# Square = indistinguishable
# ========================================================

for N, color in N_colors.items():

    # ----------------------------------------------------
    # Distinguishable battery
    # ----------------------------------------------------

    idx1 = N_arr1 == N

    plt.scatter(
        x1[idx1],
        y1[idx1],
        color=color,
        marker='o',
        s=90,
        edgecolors='k',
        linewidth=0.6
    )


    # ----------------------------------------------------
    # Indistinguishable battery
    # ----------------------------------------------------

    idx2 = N_arr2 == N

    plt.scatter(
        x2[idx2],
        y2[idx2],
        color=color,
        marker='s',
        s=90,
        edgecolors='k',
        linewidth=0.6
    )


# ========================================================
# Theoretical upper bounds
# ========================================================

plt.plot(
    xb,
    bound_finite,
    color='black',
    linestyle='-',
    linewidth=2,
    label='General upper bound'
)

plt.plot(
    xb,
    bound_largeN,
    color='black',
    linestyle='--',
    linewidth=2,
    label='Positive-work upper bound'
)


# ========================================================
# Same x and y axis limits for the entire figure
# ========================================================

plt.xlim(0.62, 0.88)
plt.ylim(-0.05, 2.5)


# ========================================================
# Axis labels
# ========================================================

plt.xlabel(
    r'$\mathcal{E}_b/E_{\max}$',
    fontsize=14
)

plt.ylabel(
    r'$\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle$',
    fontsize=14
)


# ========================================================
# Title
# ========================================================

plt.title(
    'Relative ergotropy fluctuations and theoretical '
    'upper bounds of Central Spin model',
    fontsize=12
)


# ========================================================
# Legend 1: Battery type
# ========================================================

battery_legend = [

    Line2D(
        [0],
        [0],
        marker='o',
        linestyle='None',
        color='gray',
        markeredgecolor='black',
        markersize=8,
        label='Distinguishable battery'
    ),

    Line2D(
        [0],
        [0],
        marker='s',
        linestyle='None',
        color='gray',
        markeredgecolor='black',
        markersize=8,
        label='Indistinguishable battery'
    )
]


legend1 = plt.legend(
    handles=battery_legend,
    loc='upper left',
    fontsize=9,
    title='Battery type'
)

plt.gca().add_artist(legend1)


# ========================================================
# Legend 2: N values + theoretical bounds
# ========================================================

N_legend = [
    Line2D(
        [0],
        [0],
        marker='o',
        linestyle='None',
        color=color,
        markeredgecolor='black',
        markersize=8,
        label=f'$N={N}$'
    )
    for N, color in N_colors.items()
]

bound_legend = [
    Line2D(
        [0],
        [0],
        color='black',
        linestyle='-',
        linewidth=2,
        label='General upper bound'
    ),

    Line2D(
        [0],
        [0],
        color='black',
        linestyle='--',
        linewidth=2,
        label='Positive-work upper bound'
    )
]

# Combine both
combined_legend = N_legend + bound_legend

plt.legend(
    handles=combined_legend,
    loc='best',
    fontsize=9,
    title='System size and theoretical bounds'
)


# ========================================================
# Layout
# ========================================================

plt.tight_layout()

plt.show()
```

```python
##############################################################
# Relative ergotropy fluctuations vs N with theoretical bounds
##############################################################

import numpy as np
import matplotlib.pyplot as plt


# --------------------------------------------------------
# Normalized ergotropy: E_b / (N * omega)
# --------------------------------------------------------

x1 = E_ergo1 / E_max1
x2 = E_ergo2 / E_max2


# --------------------------------------------------------
# Sort data according to N
# --------------------------------------------------------

i1 = np.argsort(N_arr1)
i2 = np.argsort(N_arr2)

N1 = N_arr1[i1]
F1 = F_ratio1[i1]

N2 = N_arr2[i2]
F2 = F_ratio2[i2]


# --------------------------------------------------------
# Theoretical bounds
# --------------------------------------------------------

x_bound = x1[i1]

general_bound = np.sqrt(1 / x_bound**2 - 1)

positive_work_bound = np.sqrt(1 / x_bound - 1)


# --------------------------------------------------------
# Plot
# --------------------------------------------------------

fig, ax = plt.subplots(figsize=(7, 5))


# Distinguishable
ax.plot(
    N1,
    F1,
    'o-',
    ms=3.5,
    lw=1.5,
    label='Distinguishable battery'
)


# Indistinguishable
ax.plot(
    N2,
    F2,
    's--',
    ms=3.5,
    lw=1.5,
    label='Indistinguishable battery'
)


# General upper bound
ax.plot(
    N1,
    general_bound,
    'k-',
    lw=1.5,
    label='General upper bound'
)


# Positive-work upper bound
ax.plot(
    N1,
    positive_work_bound,
    'k--',
    lw=1.5,
    label='Positive-work upper bound'
)


# --------------------------------------------------------
# Axes
# --------------------------------------------------------

ax.set_xlim(N1.min(), N1.max())

ax.set_xlabel(
    r'$N$',
    fontsize=13
)

ax.set_ylabel(
    r'$\sigma_{\mathcal{E}_b}/\langle\mathcal{E}_b\rangle$',
    fontsize=13
)

# --------------------------------------------------------
# Title
# --------------------------------------------------------

plt.title(
    'Relative ergotropy fluctuations vs system size of Central Spin model',
    fontsize=12
)

# --------------------------------------------------------
# Compact style
# --------------------------------------------------------

ax.tick_params(
    axis='both',
    labelsize=11
)


ax.legend(
    fontsize=9,
    frameon=True,
    loc='best'
)

plt.tight_layout()
plt.show()
```
